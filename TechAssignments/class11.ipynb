Write the following functions:

   Return: Strike rate rounded to 2 decimals

2. calculate_batting_average(player_dict)
   Formula: Total runs / Number of innings
   Return: Average rounded to 2 decimals

Run functions and print the output

```python
def calculate_strike_rate(player_dict):
    playersAvg = {}
    for pid , v in player_dict.items():
         
        if "runs_scored" in v and "balls_faced" in v:
            name = v["name"]
            runs = sum(v["runs_scored"])
            balls = sum(v["balls_faced"]) 
            sr = round(((runs/balls)*100),2) 
         
            playersAvg[pid] = {
            "name": v["name"],
            "strike_rate": sr  }
        else:
            continue
        

    return playersAvg



strikeRate = calculate_strike_rate(players)
strikeRate         

    
```

TASK 2.2: Performance Trends
1. Calculate 3-match moving average for each player's runs
2. Identify matches where player scored > mean + 1 std (exceptional performances)
3. Find correlation between balls_faced and runs_scored for all batsmen

```python
import numpy as np
player_stats = {}
for pid , data in players.items():

    if "runs_scored" in data:
        runs = np.array(data["runs_scored"])
        mean = np.mean(runs)
        std  = np.std(runs)
        med  = np.median(runs)
        high = np.max(runs)
        low  = np.min(runs)

        if std != 0:
            cs   = round((mean / std),2)
        else:
            cs  = 0

        player_stats[pid] = {
        "name": data["name"],
        "mean": round(mean, 2),
        "std": round(std, 2),
        "median": med,
        "highest": high,
        "lowest": low,
        "consistency": cs
        } 


player_stats

```
