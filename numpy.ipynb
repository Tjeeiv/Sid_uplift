```python
np.array([1, 2, 3])            # int64   by default
np.array([1.0, 2.0, 3.0])      # float64 by default
np.array([1, 2, 3], dtype=float)  # force float
np.array([1, 2, 3], dtype=str)    # force string
```

```python
# Python list — slow, one by one
a = [1, 2, 3, 4]
b = [x * 2 for x in a]   # need a loop

# NumPy array — fast, all at once
a = np.array([1, 2, 3, 4])
b = a * 2                 # [2 4 6 8] ← no loop needed!

#NumPy operates on the entire array at once — this is called vectorization
```

```python
b = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print(b[1,2])
print(b[:,2])
print(b[0:2, 0:2])
print(b[b>45]) 
print(b[::-1])
```

```python
# You're cleaning customer data
customers = np.array([100, 200, 300, 400, 500])

# You want to test something on first 3 only
test = customers[0:3]   # ← VIEW!
test[0] = 0             # ← accidentally wipes original data! 😱

# Safe way
test = customers[0:3].copy()   # ← COPY!
test[0] = 0                    # ← original customers safe ✅
```
