```python
import os 
from openai import OpenAI

apikey = os.getenv("Puter_API_KEY")
```

```python
print(apikey)
```

```python

client = OpenAI(base_url= "https://api.puter.com/puterai/openai/v1/" , api_key=apikey)
```

```python
response = client.chat.completions.create(model="gpt-4o", messages=[{"role": "user", "content":"Hi!"}])

print(response.choices[0].message.content)
```

```python
print(response )
```

```python
def get_completion(prompt , Model = "gpt-4o", system_prompt = "You are a helpful assistant."):
    messages=[
          {"role":"system" , "content":system_prompt} , {"role":"user" , "content":prompt}
     ]
    response =  client.chat.completions.create(model = Model ,messages= messages, temperature= 0.7 )

    return response.choices[0].message.content
```

```python
print(get_completion("Hi!"))
```

```python
response =  client.chat.completions.create(model = "gpt-4o" ,messages=[{"role": "user", "content":"give me paragraph of 100 words based on digital ghost"}],
                                           stream = True) 

for chunks in response:
    content = chunks.choices[0].message.content
    if content:
        print(content , end="",flush=True)
```
